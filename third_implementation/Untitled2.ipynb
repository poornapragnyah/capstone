```python
from sentence_transformers import SentenceTransformer, util
import cv2
import numpy as np

model = SentenceTransformer("clip-ViT-L-14")

def video_embedding(path):
    cap = cv2.VideoCapture(path)
    embs = []
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        emb = model.encode(frame_rgb, convert_to_tensor=True)
        embs.append(emb)
    cap.release()
    return np.mean(embs, axis=0)

emb1 = video_embedding("/home/poorna/Downloads/dataa/1st_10min/clip_004.mp4")
emb2 = video_embedding("./output_videos_gemini/idx_4.mp4")

similarity = util.cos_sim(emb1, emb2)
print(similarity)
```
